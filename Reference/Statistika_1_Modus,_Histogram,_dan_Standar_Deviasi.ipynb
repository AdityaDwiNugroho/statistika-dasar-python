**Standar Deviasi**

```python
intervals = [(20, 29), (30,39),(40,49),(50,59),(60,69),(70,79),(80,89)]
frekuensi = [3, 6, 12, 15, 12, 9, 3]


# 1. Buat DataFrame
df = pd.DataFrame({
    "Intervals": intervals,
    "Frekuensi": frekuensi
})

df
```

```python
# 2. Midpoint

df["Midpoint"] = df["Intervals"].apply(lambda x: (x[0] + x[1])/2)

df
```

```python
# 3. Total Frekuensi
N = df["Frekuensi"].sum()

print(N)
```

```python
# 4. Mean

mean = (df["Frekuensi"] * df["Midpoint"]).sum() / N
```

```python
# 5. Varians

varians = ((df["Midpoint"] - mean) ** 2 * df["Frekuensi"]).sum() / (N - 1)
```

```python
# 6. Standar Deviasi
std = np.sqrt(varians)
```

```python
# 7. Deviasi Rata-rata
deviasi_rata2 = std / np.sqrt(N)
```

```python
# Output
print("Total Frekuensi: ", N)
print("Mean: ", mean)
print("Varians: ", varians)
print("Standar Deviasi: ", std)
print("Deviasi Rata-rata: ", deviasi_rata2)
```
