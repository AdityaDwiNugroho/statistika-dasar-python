### Perhitungan Nilai Dispersi

```python
min_val5252 = Data5252['Nilai5252'].min()
max_val5252 = Data5252['Nilai5252'].max()
range_val5252 = max_val5252 - min_val5252

varians5252 = Data5252['Nilai5252'].var()
std_dev5252 = Data5252['Nilai5252'].std()

q1_5252 = Data5252['Nilai5252'].quantile(0.25)
q2_5252 = Data5252['Nilai5252'].quantile(0.50)
q3_5252 = Data5252['Nilai5252'].quantile(0.75)
iqr5252 = q3_5252 - q1_5252

print("--- Hasil Nilai Dispersi ---")
print("1. Jangkauan (Range):", range_val5252)
print("2. Varians (Variance):", varians5252)
print("3. Standar Deviasi (Standard Deviation):", std_dev5252)
print("4. Kuartil 1 (Q1):", q1_5252)
print("5. Kuartil 2 (Q2/Median):", q2_5252)
print("6. Kuartil 3 (Q3):", q3_5252)
print("7. Jangkauan Antar Kuartil (IQR):", iqr5252)
```
