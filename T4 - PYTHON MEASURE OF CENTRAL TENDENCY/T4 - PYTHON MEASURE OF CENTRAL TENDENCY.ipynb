### Import Library & Dataset

```python
import os
import pandas as pd5252
import matplotlib.pyplot as plt5252

file_name5252 = 'Data Statistika Nilai 40 siswa.csv'
if os.path.exists(file_name5252):
    Data5252 = pd5252.read_csv(file_name5252, header=None, names=['Nilai5252'])
    print("Dataset berhasil diimport!")
else:
    print(f"File {file_name5252} tidak ditemukan!")
```

### Tampilkan Dataset

```python
Data5252
```

### Hasil Analisis Deskriptif

```python
mean5252 = Data5252['Nilai5252'].mean()
median5252 = Data5252['Nilai5252'].median()
modus5252 = Data5252['Nilai5252'].mode()[0]

print("Mean of the dataset:")
print("Mean:", mean5252)
print("\nMedian of the dataset:")
print("Median:", median5252)
print("\nMode of the dataset:")
print("Mode:", modus5252)
```

### Visualisasi Diagram Mean

```python
plt5252.figure(figsize=(6, 4))
plt5252.hist(Data5252['Nilai5252'], bins=10, color='skyblue', edgecolor='black')
plt5252.axvline(mean5252, color='red', linestyle='dashed', linewidth=2, label=f'Mean ({mean5252:.2f})')
plt5252.title('Histogram dengan Letak Mean')
plt5252.xlabel('Nilai')
plt5252.ylabel('Frekuensi')
plt5252.legend()
plt5252.show()
```

### Visualisasi Diagram Median

```python
plt5252.figure(figsize=(6, 4))
plt5252.hist(Data5252['Nilai5252'], bins=10, color='skyblue', edgecolor='black')
plt5252.axvline(median5252, color='blue', linestyle='dashed', linewidth=2, label=f'Median ({median5252:.2f})')
plt5252.title('Histogram dengan Letak Median')
plt5252.xlabel('Nilai')
plt5252.ylabel('Frekuensi')
plt5252.legend()
plt5252.show()
```

### Visualisasi Diagram Modus

```python
plt5252.figure(figsize=(6, 4))
plt5252.hist(Data5252['Nilai5252'], bins=10, color='skyblue', edgecolor='black')
plt5252.axvline(modus5252, color='green', linestyle='dashed', linewidth=2, label=f'Modus ({modus5252:.2f})')
plt5252.title('Histogram dengan Letak Modus')
plt5252.xlabel('Nilai')
plt5252.ylabel('Frekuensi')
plt5252.legend()
plt5252.show()
```
